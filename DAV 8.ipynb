```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import(
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
data = {
    "Study_Hours":[1,2,3,4,5,6],
    "Attendance":[50,60,65,70,80,85],
    "Result":[
        "Fail", "Fail", "Fail",
        "Pass", "Pass", "Pass"
    ]
}
df = pd.DataFrame(data)
print("Student Dataset")
```

```python
x=df[["Study_Hours","Attendance"]]
y=df["Result"] 
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
print("\nTraining Records")
print(len(x_train))
```

```python
print("\nTesting Records")
print(len(x_test))
```

```python
model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=2,
    random_state=42
)
```

```python
model.fit(x_train, y_train)
print("\nDecision Tree Model Trained Successfully")
```

```python
plt.figure(figsize=(16, 9), dpi=150)
plot_tree(
    model,
    feature_names=["Study_Hours", "Attendance"],
    class_names=["Fail", "Pass"],
    filled=True,
    rounded=True,
    proportion=False,
    precision=2,
    fontsize=12
)
```

```python
plt.title("Decision Tree - Student Pass/Fail Prediction",fontsize=16)
plt.tight_layout()
plt.show()
```

```python
y_pred = model.predict(x_test)
print("\nActual Results:")
print(y_test.values)
print("\npredicted results")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("\nModel Accuracy:",accuracy*100,"%")
```

```python
cm =confusion_matrix(
    y_test,
    y_pred,
    labels=["Fail","Pass"]
)
print(cm)
```

```python
plt.figure(figsize=(6,4))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Fail", "Pass"],
    yticklabels=["Fail", "Pass"]
)

plt.xlabel("predicted result")
plt.ylabel("actual result")
plt.title("confusion matrix")
plt.tight_layout()
plt.show()
```

```python
# Step 11: Classification Report
print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        labels=[0, 1],
        target_names=["Fail", "Pass"],
        zero_division=0
    )
)

# Step 12: Predict Result for a New Student
new_student = pd.DataFrame({
    "Study_Hours": [4],
    "Attendance": [85]
})

prediction = model.predict(new_student)

print("\nNew Student Details:")
print(new_student)

print("\nPredicted Student Result:", prediction[0])
```
